```python
import os

# Base directory
base_dir = "child-speech-ai"

# List of folders to create
folders = [
    "data/raw",
    "data/processed",
    "data/external",
    "notebooks",
    "src/data",
    "src/models",
    "src/evaluation",
    "src/utils",
    "models",
    "deployment/mobile_app",
    "deployment/teacher_dashboard",
    "deployment/ondevice_inference",
    "tests",
    "docs"
]

# Create folders
for folder in folders:
    path = os.path.join(base_dir, folder)
    os.makedirs(path, exist_ok=True)

# Create placeholder files
files = [
    "requirements.txt",
    "environment.yml",
    "README.md",
    "LICENSE",
    ".gitignore",
    "docs/methodology.md",
    "docs/usage_guide.md"
]

for file in files:
    path = os.path.join(base_dir, file)
    with open(path, "w") as f:
        f.write("")  # empty file for now

print(f"Repository structure created under '{base_dir}/'")
```
