```python
## Decision Tree Regression
# Importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Importing dataset
dataset = pd.read_csv("/content/drive/MyDrive/task3_decision_tree_data.csv")
X = dataset.iloc[:, 0:1].values
Y = dataset.iloc[:, 1:2].values

```

```python
# Fitting Decision Tree Regression to dataset
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(random_state=0)
regressor.fit(X, Y)
```

```python
# Visualising Decision Tree Regression results
plt.scatter(X, Y)
plt.plot(X, regressor.predict(X), color='red')
plt.title('Exam Dataset(Decision Tree)')
plt.xlabel('Hours Studied')
plt.ylabel('Exam score')
plt.show()
```

```python
# Predicting results
Y_pred = regressor.predict([[8.5]])
```

```python
# Visualising Decision Tree Regression results (Higher resolution)
X_grid = np.arange(min(X), max(X), 0.01)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, Y)
plt.plot(X_grid, regressor.predict(X_grid), color='red')
plt.title('Exam data (Decision Tree)')
plt.xlabel('Hours studied')
plt.ylabel('Exam_Score')
plt.show()
```

```python
#Building a linear regression
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(X, Y)
```

```python
#visualizing linear regression results
plt.scatter(X,Y)
plt.plot(X, lin_reg.predict(X), color='red')
plt.title('Exam Dataset (Linear Regression)')
plt.xlabel('Hours studied')
plt.ylabel('Exam_Score')
plt.show()
```

# Comparing Decision Tree vs Linear Regression predictions
1. Linear Regression: The "Trend Follower"
Linear regression looks at the "slope" of my data. It notices that as hours increase, scores go up at a steady rate. Even though 23 hours isn't in our dataset, the model follows the math of its straight line to calculate a predicted score of 75.34. It "guesses" what comes next based on the pattern.

2. Decision Tree: Also known as The "Box Maker"
A Decision Tree doesn't use a continuous line. Instead, it splits our data into boxes or groups.
It looks at our data and sees the highest category in the dataset provided was 22 hours (Score: 75).
When we ask about 23 hours, the model says: "This belongs in my highest category."
It simply returns the average score of that category, which is 75.0. It cannot predict anything higher than the highest value it has already seen.
Summary:
Linear Regression is mathematical and predicts a trend (resulting in decimals).
Decision Tree is logic-based and predicts a category (resulting in "stair-step" values).

## The advantages of using the decision tree regressor:
Here are the advantages of using a Decision Tree for study and exam data:

1. Handling "Non-Linear" Breaks
In real life, studying doesn't always give a steady increase. You might hit a "wall" where studying from hour 20 to 22 doesn't actually improve your score because of fatigue.

Linear Regression would still try to draw a straight line through that fatigue zone.

Decision Trees can recognize that "Hours > 20" results in a flat score, creating a distinct "step" in the data.

2. Automatic Feature Selection
If we were to add more faetures to the dataset like "Coffee Consumed," "Hours of Sleep," and "Room Temperature," a Decision Tree is excellent at figuring out which one matters most.

It places the most important variable (likely Hours Studied) at the very top of the tree.

It ignores "noisy" or useless data automatically, whereas Linear Regression can be easily confused by irrelevant columns.

3. No Need for "Feature Scaling"
Recalling the previous discussion about SVR and Polynomial Regression requiring StandardScaler.

Decision Trees do not care about the scale of your numbers.

Whether you measure study time in minutes (1,200) or hours (20), the tree will make the exact same "Yes/No" splits. This makes your code much simpler and less prone to errors.

4. Capturing "Interaction" Between Variables
A Decision Tree is great at finding conditional relationships. For example:

If Hours Studied > 10 AND Sleep < 4 hours, then Score = 60.

If Hours Studied > 10 AND Sleep > 8 hours, then Score = 90. It "branches" based on combinations of factors in a way that feels very much like human logic.
