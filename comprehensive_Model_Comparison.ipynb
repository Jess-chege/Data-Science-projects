````python
#comprehensive Model Comparison between linear regression, decision tree, and svr
# Importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

````

````python
# Importing dataset
dataset = pd.read_csv("/content/drive/MyDrive/assignment1_salary_prediction.csv")
X = dataset.iloc[:, 0:1].values
Y = dataset.iloc[:, 1:2].values
````

````python
#creating linear regression
from sklearn.linear_model import LinearRegression #importing linear regression model from the sklearn library

#creating the Linear regression model
lin_reg = LinearRegression()#calling the model
lin_reg.fit(X, Y)#fitting the dataset to the model
````

````python
#Predict salary for position level 6.5
pred_lin = lin_reg.predict([[6.5]])
print("Linear Regression Prediction (6.5 Level):", pred_lin[0])
````

````python
#creating the Polynormial model degree 2 which is used to transform the input data
#importing the Polynormial model
from sklearn.preprocessing import PolynomialFeatures

pol_reg = PolynomialFeatures(degree=2)#transforming tool
X_poly= pol_reg.fit_transform(X)
````

````python
#creating linear regression model 2 and fitting in the poly transformed
lin_reg2 = LinearRegression()
lin_reg2.fit(X_poly, Y)
````

````python
#creating the Polynormial model degree 3 which is used to transform the input data
pol_reg3 = PolynomialFeatures(degree=3)#transforming tool
X_poly3= pol_reg3.fit_transform(X)
````

````python
#creating linear regression model 3 and fitting in the poly transformed
lin_reg3 = LinearRegression()
lin_reg3.fit(X_poly3, Y)
````

````python
#creating the Polynormial model degree 4 which is used to transform the input data
pol_reg4 = PolynomialFeatures(degree=4)#transforming tool
X_poly4= pol_reg4.fit_transform(X)
````

````python
#creating linear regression model 4 and fitting in the poly transformed
lin_reg4 = LinearRegression()
lin_reg4.fit(X_poly4, Y)
````

````python
#Creating Visualisations

#Regression line/curve
#Proper title and labels

X_grid = np.linspace(X.min(), X.max(), 100).reshape(-1,1)  # smoother curves for poly models

# Linear Regression
#1.Original data points
plt.scatter(X, Y, color='green', label='Original Data')
plt.plot(X, lin_reg.predict(X), color='red', label='Linear Regression')
plt.title('Linear Regression')
plt.xlabel('Position_Level')
plt.ylabel('Salary')
plt.legend()
plt.show()
````

````python
#creating the SVR model with standard scaler
#feature scaling
#anytime one is using the svr model one must do the feature scaling using the standard scaler
from sklearn.preprocessing import StandardScaler
sc_Y = StandardScaler()
sc_X = StandardScaler()
X_scaled = sc_X.fit_transform(X)
Y_scaled= sc_Y.fit_transform(Y)
````

````python
#Fitting the SVR Dataset
from sklearn.svm import SVR
regressor= SVR(kernel='rbf')
regressor.fit(X_scaled,Y_scaled)
````

````python
#Creating visualizations to show how the SVR results appear
plt.scatter(X_scaled,Y_scaled)
plt.plot(X_scaled, regressor.predict(X_scaled), color = 'green')
plt.title('Salary prediction (SVR Model)')
plt.xlabel('position_level')
plt.ylabel('Salary')
plt.show()
````

````python
#predicting the 6.5 using the SVR model
#Predict salary for position level 6.5
Y_pred= regressor.predict([[6.5]])
print(Y_pred)
````

````python
# Fitting Decision Tree Regression to dataset
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor(random_state=0)
regressor.fit(X, Y)
````

````python
# Visualising Decision Tree Regression results
plt.scatter(X, Y)
plt.plot(X, regressor.predict(X), color='red')
plt.title('Salary dataset(Decision Tree)')
plt.xlabel('Position_Level')
plt.ylabel('Salary')
plt.show()
````

````python
#Predict salary for position level 6.5

#Predicting exam score for 23 hours of study using Decision tree regressor
pred = regressor.predict([[6.5]])
print(pred)
````
