```python
#Math Quiz
#Build a program that asks the user 5 random math questions (addition, subtraction, multiplication, division) and keeps score of correct answers.
import math
import random
score = 0
print("welcome to our maths quiz lets get started")

for i in range(5):
  operator =  random.choice(['+', '-', '*', '/'])
  a = random.randint(1, 20)
  b = random.randint(1, 20)

  if operator == '/':
    a = a / b
    correct_answer = round(a / b)
  elif operator == '+':
    correct_answer = a + b
  elif operator == '-':
    correct_answer = a - b
  elif operator == '*':
    correct_answer = a * b

#having the user input
user_input = input(f"Quiz1{i+1}: What is {a} {operator} {b}? ")
try:
  user_answer = float(user_input)
  if operator == '/' and math.isclose(user_answer, correct_answer, abs_tol=0.01):
    print("Correct!")
    score += 1
  elif user_answer == correct_answer:
    print("Correct!")
    score += 1
  else:
    print(f"Incorrect. The correct answer is {correct_answer}")
except ValueError:
    print(" Invalid input.")

print(f"\nYour final score is: {score}/5")




```

```python
#Date Calculator
#Write a program that takes a user’s birthday as input and tells them how many days are left until their next birthday.
import datetime

user_input = input("Enter your birthday date (day, month, year): ")

#using the Split input and convert to integers
day, month, year = map(int, user_input.split(','))

# Create a datetime object for birthday
birthday = datetime.datetime(year, month, day)

# Geting current datetime
now = datetime.datetime.now()

birthday_days = birthday - now

print(f"You have  {birthday_days.days} days remaining to your next birthday.")
```
