What is the total CIF value by commodity (groupby cmdCode)?


```python
#What is the total CIF value by commodity (groupby cmdCode)
cif_by_commodity = data.groupby('cmdCode')['cifvalue'].sum().reset_index().sort_values('cifvalue', ascending=False)
#adding commas to cif values
cif_by_commodity['cifvalue'] = cif_by_commodity['cifvalue'].apply(lambda x: f"{x:,.2f}")
cif_by_commodity.head(2)
```

Total CIF Value by Commodity

I grouped the dataset by cmdCode to calculate the total CIF value for each commodity. By summing and sorting the CIF values, I identified which commodities contribute most to imports. I also formatted the values with commas for readability.

In this dataset, the two commodities are:

Coffee — with a total CIF value of 77,414,571.52

Avocados — with a total CIF value of 50,825,998.31

This shows that coffee contributes the most to import value, followed by avocados. Grouping by commodity makes it easy to see the relative importance of each product in the country’s import profile.

Trade flow

How does CIF value change over time (groupby refYear or refMonth)?

```python
# Total CIF value per year
cif_by_year = data.groupby('refYear')['cifvalue'].sum().reset_index()

# Format with commas for readability
cif_by_year['cifvalue'] = cif_by_year['cifvalue'].apply(lambda x: f"{x:,.0f}")

cif_by_year

```

1.Peak Import Periods:

The highest CIF values occur in months 4, 5, and 7, indicating that imports were concentrated during these months.

2.Low Import Periods:

The lowest CIF values are in months 10, 11, and 12, suggesting seasonal slowdowns.

3.Seasonal Pattern:

The pattern shows a rise in imports during early to mid-year, followed by a steady decline toward the end of the year.

This might reflect seasonal production or demand cycles (e.g., avocado harvest periods or coffee exports).

```python
#grouping the dataset into imports and exports
flow_groups = data.groupby('flowCode')
#separating the imports and the exports
imports = flow_groups.get_group('Import')
exports = flow_groups.get_group('Export')
```

```python
# Monthly total for imports
monthly_imports = imports.groupby(['refYear', 'refMonth'])['cifvalue'].sum().reset_index()
monthly_imports['cifvalue'] = monthly_imports['cifvalue'].apply(lambda x: f"{x:,.0f}")
monthly_imports.sort_values('cifvalue', ascending=False, inplace=True)
monthly_imports.head(5)  # Top 5 months by imports

```

```python
# Monthly total for exports
monthly_exports = exports.groupby(['refYear', 'refMonth'])['cifvalue'].sum().reset_index()
monthly_exports['cifvalue'] = monthly_exports['cifvalue'].apply(lambda x: f"{x:,.0f}")
monthly_exports.sort_values('cifvalue', ascending=False, inplace=True)
monthly_exports.head(5)  # Top 5 months by exports
```

```python
import matplotlib.pyplot as plt
```

```python
#grouping by seasons
#months and cif values picked from monthly_exports for plotting
months = ['1-2', '8-9', '0-1', '10-11', '9-10']  # Year-Month labels
cif_values = [7354654, 6831049, 5334198, 3339343, 3163177]  # CIF values as numbers

# Plot bar graph
plt.figure(figsize=(8,5))
plt.bar(months, cif_values, color='skyblue')
plt.title('Top 5 Months by CIF Value')
plt.xlabel('Year-Month')
plt.ylabel('CIF Value')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#plotting for the  3 importers/exporters of a commodity using groupby
data_dict = {
    'partnerCode': ['USA', 'NLD', 'CHN'],
    'cifvalue': [6.008638e+07, 5.452706e+07, 1.362712e+07]
}

df = pd.DataFrame(data_dict)

# Plot bar chart
plt.figure(figsize=(8,5))
plt.bar(df['partnerCode'], df['cifvalue'], color='skyblue')
plt.title('Top 3 Importers by CIF Value')
plt.xlabel('Partner Country')
plt.ylabel('CIF Value')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#plot a bar chart showing the top three importers or exporters of your selected trade item over the period you grabbed the data for, compared to the total world trade value.
#saving the grouped data in a dictionary
data_dict = {
    'partnerCode': ['USA', 'NLD', 'CHN'],
    'cifvalue': [6.008638e+07, 5.452706e+07, 1.362712e+07]
}
df_top3 = pd.DataFrame(data_dict)

# Total world trade value for this commodity
world_total = df_top3['cifvalue'].sum() + 2.0e+07  # Example: adding remaining world value

# Plot
plt.figure(figsize=(8,5))
bars = plt.bar(df_top3['partnerCode'], df_top3['cifvalue'], color='skyblue', label='Top 3 Importers')

# Add a horizontal line for world total
plt.axhline(world_total, color='green', linestyle='--', linewidth=2, label='World Total')

plt.title('Top 3 Importers vs World Total')
plt.xlabel('Partner Country')
plt.ylabel('CIF Value')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

The analysis shows that a small number of countries dominate the import market for this commodity, with the top three importers (USA, NLD, and CHN) accounting for a significant portion of the total world trade value. While many other countries also import this commodity, their combined contribution is relatively minor, indicating a high concentration of trade among a few key players. This suggests that global trade flows for this commodity are heavily dependent on these major importers, and changes in their demand or policies could have a substantial impact on the overall market.
