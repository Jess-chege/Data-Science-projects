```python
#importing libraries
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url= "https://www.worldometers.info/world-population/"
```

```python
data = pd.read_html(url, header=0)
```

```python
print(data)
```

```python
#countries with higher population
chp= data[4]
```

```python
chp.head(10)
```

```python
#renaming the population column from population 2025 to population
chp_new=chp.rename(columns={'Population 2025': 'Population'})
chp_new.head(10)
```

```python
#plotting a line graph using the yearly change and country or dependancy
fig, ax = plt.subplots(figsize=(5, 6))

chp_new.plot(ax=ax, linewidth=8)
ax.set_xlabel('Yearly Change')
ax.set_ylabel('Country (or dependency)')

plt.show()
```

```python
#future world data population
predicted=data[1]
```

```python
predicted.head(10)
```

```python
#converting the yearly % chamge column from string to float
predicted['Yearly % Change'] = (
    predicted['Yearly % Change']
    .str.replace('%', '', regex=False)
    .astype(float)
)
```

```python
#plotting a bar graph using the years and population change in the future

fig, ax = plt.subplots(figsize=(11, 11))

ax.plot(predicted['Year (July 1)'], predicted['Population'], linewidth=4)
ax.set_ylabel('Population')


plt.show()
```

```python
#combining the two datasets in one line plot
fig, ax = plt.subplots(figsize= (10, 10))
chp_new.plot(ax=ax, linewidth=4)
predicted.plot(ax=ax, linestyle='dashed', linewidth=4)
ax.set_ylabel('Global Population')
plt.show()
```

```python
#concatinating the two datasets
all_data = pd.concat([chp_new, predicted], axis = 1)
```

```python
all_data
```

### Population Summary

This dataset combines country level population data for 2025 with global projections up to 2050. It includes population numbers, growth rates, fertility, median age, density, and urbanization.

### Key Points:

Top countries: India (1.46B), China (1.42B), and the US (347M) are the most populated, making up a big chunk of the world population.

Global growth: The world population goes from 8.57B in 2030 to 9.66B in 2050, but growth is slowing (0.81% :  0.47%).

Aging population: Median age rises from 32 in 2030 to 36 by 2050. Fertility rates drop slightly, which is slowing population growth.

Density & urbanization: India and Pakistan are very dense, while small countries like Montserrat or Niue have tiny populations. China and the US are highly urbanized.

Takeaway: Population is still growing, but unevenly. Big countries drive global numbers, and aging plus slowing growth are key trends to watch.

```python
# Checking number of rows in each table
for i, table in enumerate(tables):
    print(f"Table {i} has {table.shape[0]} rows and {table.shape[1]} columns")

# Selected the table with 233 rows
df = tables[0]  # usually the first one
print(df.shape)
```

```python
#selecting the 4th table to use in my plotting
df1= tables[4]
df1.head(10)
```

```python
#plotting data from the largest table
top10 = df1.nlargest(10, 'Population 2025')

plt.figure(figsize=(12,6))
plt.bar(top10['Country (or dependency)'], top10['Population 2025'], color='skyblue')
plt.xlabel('Country')
plt.ylabel('Population')
plt.title('Top 10 Most Populated Countries (2025)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Distribution of Yearly Population Growth

plt.figure(figsize=(12,6))
plt.hist(df1['Yearly Change'], bins=20, color='orange', edgecolor='black')
plt.xlabel('Yearly Change')
plt.ylabel('Number of Countries')
plt.title('Distribution of Yearly Population Growth')
plt.tight_layout()
plt.show()
```

```python
#distribution Population vs Yearly Change
plt.figure(figsize=(12,6))
plt.scatter(df1['Population 2025'], df1['Yearly Change'], color='green', alpha=0.6)
plt.xlabel('Population 2025')
plt.ylabel('Yearly Change')
plt.title('Population vs Yearly Growth')
plt.tight_layout()
plt.show()
```

## Executive Summary: Global Population Insights (1955–2050)

Overview
I analyzed historical, current, and projected global population trends using multiple datasets, including:

Historical population from 1955–2025 with yearly changes, fertility rates, median ages, and density.

Projected global population from 2030–2050.

Continental and regional populations for 2025.

Country-level population data for 2025 (233 countries).

Key Insights

### Global Population Growth

The world population grew from 2.74 billion in 1955 to 8.23 billion in 2025.

By 2050, projections show it reaching 9.66 billion.

Growth rates are slowing: from 1.91% in 1955 down to 0.85% in 2025, dropping further to 0.47% by 2050.

### Demographics

Median age is rising from 21.8 years in 1955 to 30.9 in 2025, showing that populations are aging.

Fertility rates are dropping globally from 5.0 children per woman in 1955 to 2.24 in 2025 explaining slower growth.

## Regional Population

Asia: 4.83 billion (58.7% of global population)

Africa: 1.55 billion (18.8%)

Europe: 744 million (9%)

Latin America & Caribbean: 668 million (8.1%)

North America: 388 million (4.7%)

Oceania: 46 million (0.57%)

### Top Countries (2025)
India: 1.46 billion (0.89% yearly growth)
China: 1.42 billion (slightly declining at −0.23%)
USA: 347 million (0.54%)
Others in top 10: Indonesia, Pakistan, Nigeria, Brazil, Bangladesh, Russia, Ethiopia

### Population Density
Asia and Europe are crowded, which impacts urban planning and resources.
Africa has lower density but higher growth—cities and infrastructure will need attention soon.

### What This Means

Policy: Aging populations in Asia and Europe will need healthcare and workforce planning. High-growth regions like Africa need infrastructure, schools, and hospitals.
Business: Fast-growing markets like India and Nigeria are opportunities for consumer products and services.
Sustainability: Population density and urbanization trends mean we need smart city planning and resource management.

### Bottom Line
The global population is still growing but slower. Growth is uneven: some regions are booming while others are stabilizing or even declining. Understanding these trends is crucial for planning, investment, and policy decisions over the next 30 years.
