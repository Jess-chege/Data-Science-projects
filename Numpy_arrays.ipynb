```python
#Write a NumPy program to test element-wise for NaN of a given array.
#Nan elements are the missing elements represented as NaN meaning not numbers in an array

import numpy as np

# Create an array with some NaN values
array = np.array([1, 3, np.nan, 7, np.nan, 10])

# Test element-wise for NaN
results = np.isnan(array)

print("Original Array:")
print(array)

print("\nElement-wise NaN test:")
print(results)


```

```python
#Write a NumPy program to test element-wise for complex number, real number of a given array. Also test whether a given number is a scalar type or not.
#complex number: 1.0 + 2.0j, 1.5 + 2.5j
#real number: -3, -7.2
#scalar types: single value type in the array
import numpy as np
#an array that contains both complex and real numbers
arr = np.array([1.0 + 2.0j, 5])

#Test element-wise for complex numbers
is_complex = np.iscomplex(arr)

#Test element-wise for real numbers
is_real = np.isreal(arr)
#testing if a given number is scalar
x = 30
is_x_scalar = np.isscalar(x)

print("Original Array:")
print(arr)
#testing if the elements in the array are complex
print("\nElement-wise complex number test:")
print(is_complex)
#testing if the elements in the array are real numbers
print("\nElement-wise real number test:")
print(is_real)
#checking if the element in array x is scalar
print(f"\nIs {x} a scalar? →", is_x_scalar)
```

```python
#Write a NumPy program to test whether two arrays are element-wise equal within a tolerance.
import numpy as np

# Create two arrays
arr1 = np.array([1.0, 3.003, 4.0005, 5.0])
arr2 = np.array([1.0, 3.009, 4.0004, 5.1])

# Test element-wise equality within a tolerance
# absolutetol = absolute tolerance, relativetol = relative tolerance
result = np.isclose(arr1, arr2, rtol=1e-03, atol=1e-05)

print("Array 1:", arr1)
print("Array 2:", arr2)
print("\nAre they element-wise equal within tolerance?")
print(result)
```

```python
#Write a NumPy program to create an element-wise comparison (greater, greater_equal, less and less_equal) of two given arrays.
#creating the two arrays
arr1 = np.array([2, 3, 143, 5, 6, 7])
arr2 = np.array([12, 23, 34, 55, 16, 18])

#printing the arrays
print(arr1)
print(type(arr1))
print(arr2)
print(type(arr2))

#comparing the two arrays using different functions
result1 = np.greater(arr1 , arr2)
print(result1)

result2 = np.greater_equal(arr1, arr2)
print(result2)

result3 = np.less(arr1, arr2)
print(result3)

result4 = np.less_equal(arr1,arr2)
print(result4)
```

```python
#Write a NumPy program to create an element-wise comparison (equal, equal within a tolerance) of two given arrays.
#creating the two arrays
import numpy as np

# Create two arrays
arr1 = np.array([1.0, 2.0001, 3.0, 4.5])
arr2 = np.array([1.0, 2.0002, 3.5, 4.5])

# Element-wise comparison for exact equality
exact_equal = np.equal(arr1, arr2)

# Element-wise comparison within a tolerance
#relative tolerance and the absolute tolerance
tolerance_equal = np.isclose(arr1, arr2, rtol=1e-03, atol=1e-05)

print("Array 1:", arr1)
print("Array 2:", arr2)

print("\nElement-wise exact equality:")
print(exact_equal)

print("\nElement-wise equality within tolerance:")
print(tolerance_equal)
```

```python
#Write a Python program to change the data type of given a column or a Series. Sample Series: Original Data Series: 0 100 1 200 2 python 3 300.12 4 400 dtype: object Change the said data type to numeric: 0 100.00 1 200.00 2 NaN 3 300.12 4 400.00 dtype: float64

sample_series = pd.Series([0, 100, 1, 200, 2, 'python', 3, 300.12, 4, 400])
print("Original Data Series:")
print(sample_series)

# Changing the data type to numeric
numeric_data = pd.to_numeric(sample_series, errors='coerce')

print("\nChange the said data type to numeric:")
print(numeric_data)
```

```python
#Write a Python Pandas program to convert the first column of a DataFrame as a Series. Sample Output: Original DataFrame col1 col2 col3 0 1 4 7 1 2 5 5 2 3 6 8 3 4 9 12 4 7 5 1 5 11 0 11 1st column as a Series: 0 1 1 2 2 3 3 4 4 7 5 11 Name: col1, dtype: int64 <class 'pandas.core.series.Series'>
data = {
    'col1': [1, 2, 3, 4, 7, 11],
    'col2': [4, 5, 6, 9, 5, 0],
    'col3': [7, 5, 8, 12, 1, 11]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Converting the first column (col1) to a Series
first_col_series = df.iloc[:, 0]

print("\n1st column as a Series:")
print(first_col_series)
print(type(first_col_series))
```

```python
#Write a Pandas program to convert a given Series to an array. Sample Output: Original Data Series: 0 100 1 200 2 python 3 300.12 4 400 dtype: object Series to an array ['100' '200' 'python' '300.12' '400']
data4 = pd.Series([100, 200, 'python', 300.12, 400])
print("Original Data Series:")
print(data4)

# Convert Series to a NumPy array
array_data = data4.to_numpy()

print("\nSeries to an array:")
print(array_data)
```

```python
#Write a Pandas program to convert Series of lists to one Series. Sample Output: Original Series of list 0 [Red, Green, White] 1 [Red, Black] 2 [Yellow] dtype: object One Series 0 Red 1 Green 2 White 3 Red 4 Black 5 Yellow dtype: object
# Original Series of lists
data5 = pd.Series([
    ['Red', 'Green', 'White'],
    ['Red', 'Black'],
    ['Yellow']
])

print("Original Series of list:")
print(data5)

# Convert Series of lists to one flattened Series
flattened = data5.explode().reset_index(drop=True)

print("\nOne Series:")
print(flattened)
```

```python
#Write a Pandas program to sort a given Series. Sample Output: Original Data Series: 0 100 1 200 2 python 3 300.12 4 400 dtype: object 0 100 1 200 3 300.12 4 400 2 python dtype: object
# Original Data Series
data = pd.Series([100, 200, 'python', 300.12, 400])
print("Original Data Series:")
print(data)

# Convert all values to strings before sorting
sorted_data = data.astype(str).sort_values(ignore_index=True)

print("\nSorted Data Series:")
print(sorted_data)
```

```python
#Write a Pandas program to add some data to an existing Series. Sample Output: Original Data Series: 0 100 1 200 2 python 3 300.12 4 400 dtype: object Data Series after adding some data: 0 100 1 200 2 python 3 300.12 4 400 0 500 1 php dtype: object
# Original Data Series
data7 = pd.Series([100, 200, 'python', 300.12, 400])
print("Original Data Series:")
print(data7)

# creating new data to be added
new_data = pd.Series([500, 'php'])

# Add using the concat function the new data to the existing Series
updated_data = pd.concat([data7, new_data], ignore_index=True)

print("\nData Series after adding some data:")
print(updated_data)
```
