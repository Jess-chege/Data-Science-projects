Task 1: K-Nearest Neighbors (KNN) Classification
Objective: Build a KNN classifier to predict customer purchase behavior.

Dataset: Task-Datasets/task1_knn_customer_data.csv

Instructions:
Import the necessary libraries (pandas, numpy, sklearn)
Load the dataset and explore its structure
Separate features (Age, Annual_Salary) and target variable (Purchased)
Split the data into training (80%) and test (20%) sets
Apply feature scaling using StandardScaler
Build a KNN classifier with n_neighbors=5 and metric='minkowski' (p=2)
Train the model and make predictions
Evaluate using confusion matrix and accuracy score

```python
#importing the necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
#importing the dataset
data= pd.read_csv('/content/drive/MyDrive/task1_knn_customer_data.csv')
X = data.iloc[:, [2,3]].values
Y = data.iloc[:, 4].values
```

```python
# Splitting the  dataset into train and test set using the 20% and 80%
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.20,random_state=0)
```

```python
# Feature Scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
# Fitting k-NN to Training set
from sklearn.neighbors import KNeighborsClassifier
classifier = KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)
classifier.fit(X_train, Y_train)
```

```python
#creating the prediction
y_pred = classifier.predict(X_test)
print(y_pred)
```

```python
#KNN outputs class probabilities as the proportion of neighbors in each class
proba = classifier.predict_proba(X_test)
print(proba)
```

```python
#CAlculating the confusion matrix to test on performamnce of our classification model(KNN) using the target varaible which is the y and the y predicted by the KNN model
from sklearn.metrics import confusion_matrix
CM = confusion_matrix(Y_test, y_pred)
print(CM)
```

```python
#claculating the model accuracy to measure how the model performed
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(Y_test, y_pred)
print(accuracy)
```

### Observation:
The model performed exceptionally well on the test data, correctly classifying all 20 samples with no misclassifications. Both true positives and true negatives are perfect, resulting in an accuracy score of 100%. While this indicates excellent predictive performance, the small size of the test set suggests caution, as such perfect results may not generalize to larger or more diverse datasets and could indicate potential overfitting. Further validation on additional data or via cross-validation is recommended to confirm the model’s robustness.

Instructions:

Build two SVM classifiers:

```python
#importing the dataset for the SVM model
df = pd.read_csv('/content/drive/MyDrive/task2_svm_classification_data.csv')
x = df.iloc[:, [0, 1]].values
y = df.iloc[:, 2].values
```

```python
#splitting the dataset into train and test using 80% and 20% sets
from sklearn.model_selection import train_test_split
x_train, x_test,y_train, y_test = train_test_split(x,y,test_size=0.20, random_state = 0)
```

```python
#feature scaling for standardizing the x (independent variables)
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
#fitting the SVM to the training set using the linear svm
from sklearn.svm import SVC
model_lin= SVC(kernel = 'linear', random_state = 0)
model_lin.fit(x_train, y_train)
```

```python
#predicting the y_test using the svc model
Y_pred1 = model_lin.predict(x_test)
print('The model using kernel as linear prediction is:', Y_pred1)
```

```python
#calculating the accuracy score
accuracy1= accuracy_score(y_test, Y_pred1)
print(accuracy1)
```

```python
#Calculating the confusion matrix
CM1 = confusion_matrix(y_test, Y_pred1)
print(CM1)
```

```python
#fitting the RBF kernel SVM
model_rbf= SVC(kernel = 'rbf', random_state = 0)
model_rbf.fit(x_train, y_train)
```

```python
#predicting the y_test using the svc model
Y_pred2 = model_rbf.predict(x_test)
print('The model using rbf as a kernel prediction is:', Y_pred2)
```

```python
#calculating the accuracy score
accuracy2= accuracy_score(y_test, Y_pred2)
print(accuracy2)
```

```python
#Calculating the confusion matrix
CM2 = confusion_matrix(y_test, Y_pred2)
print(CM2)
```

### observation
Both linear and RBF kernel SVMs achieved perfect classification on the test set, with a confusion matrix showing all predictions correct and 100% accuracy. This indicates that the dataset is linearly separable, so a linear decision boundary is sufficient. While the RBF kernel allows non-linear separation, it does not improve performance in this case.

Task 3: Naive Bayes Classification
Objective: Build a Naive Bayes classifier for email spam detection.

Dataset: Task-Datasets/task3_naive_bayes_data.csv

Instructions:
Import the necessary libraries
Load the email classification dataset
Explore the data structure and feature distributions
Separate features (Word_Count, Link_Count, Uppercase_Ratio, Special_Char_Count, Contains_Urgent) and target (Is_Spam)
Split the data into training (80%) and test (20%) sets
Apply feature scaling using StandardScaler
Build a Gaussian Naive Bayes classifier
Train the model and make predictions
Evaluate using confusion matrix and accuracy score
Expected Deliverables:
Confusion matrix
Accuracy score
Discussion on why Naive Bayes is suitable for this problem

```python
# Naive Bayes Model
# Importing dataset
dataset = pd.read_csv('/content/drive/MyDrive/task3_naive_bayes_data.csv')
dataset.head(10)
```

```python
X1 = dataset[['Email_ID', 'Word_Count', 'Link_Count', 'Uppercase_Ratio',
              'Special_Char_Count', 'Contains_Urgent']].values

Y1 = dataset['Is_Spam'].values
```

```python
# Splitting the dataset into train and test set
X1_train, X1_test, Y1_train, Y1_test = train_test_split(X1,Y1,test_size=0.20,random_state=0)
```

```python
# Feature Scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X1_train = sc.fit_transform(X1_train)
X1_test = sc.transform(X1_test)

```

```python
# Fitting Naive Bayes model to  the training set
from sklearn.naive_bayes import GaussianNB
classifier = GaussianNB()
classifier.fit(X1_train, Y1_train)
```

```python
# Predicting Test set results
Y_pred3 = classifier.predict(X1_test)
print(Y_pred3)
```

```python
# Calculating the Confusion Matrix
from sklearn.metrics import confusion_matrix
cm3= confusion_matrix(Y1_test, Y_pred3)
print(cm3)
```

### Why Naive Bayes is suitable

**Features are conditionally independent:**

In email spam detection, things like Word_Count, Link_Count, Contains_Urgent can often be treated as independent signals contributing to spam probability.

Naive Bayes assumes conditional independence, which works surprisingly well for text/email classification.

**Handles categorical and continuous features:**

Features like Contains_Urgent (binary) and Word_Count (continuous) can be handled by Gaussian or Bernoulli Naive Bayes.

**Fast and simple:**

Naive Bayes requires very little computation and performs well even on small datasets.

**Probabilistic output:**

Gives probability estimates (predict_proba) for class membership, which is useful for ranking emails by spam likelihood.

**Robust for imbalanced data:**

Even if spam is rarer than non-spam, Naive Bayes can still perform well with proper prior probabilities.
