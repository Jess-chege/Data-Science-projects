```python
# selecting the input and target features

data = hearth_df.iloc[:,:13]

target = hearth_df.iloc[:,13] 
```

```python
data.head()
```

```python
#spliting the data

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(data, target, train_size= 0.8)
```

```python
Y_train.shape
```

```python
# importing the naive bayes algorithm
from sklearn.naive_bayes import BernoulliNB
```

```python
# training the model
naive_model =  BernoulliNB()
naive_model.fit(X_train, Y_train)
```

```python
# making prediction on the test data (X_test)
pred = naive_model.predict(X_test)
```

```python
# checking the model acciracy
from sklearn.metrics import accuracy_score
accuracy_score(Y_test, pred)* 100
```
